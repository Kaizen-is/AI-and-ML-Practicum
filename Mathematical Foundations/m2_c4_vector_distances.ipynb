```python
# TASK 1

import pandas as pd
import numpy as np
import math

df = pd.read_csv('superstore_cleaned.csv')
data = {'Sales': [250.0, 240.0, 800.0], 'Quantity': [2, 3, 5], 'Discount': [0.1, 0.2, 0.0]}
df = pd.DataFrame(data)
features = ['Sales', 'Quantity', 'Discount']
products = df[features].head(3)
p1 = products.iloc[0].values
p2 = products.iloc[1].values
p3 = products.iloc[2].values

print(f"Product 1: {p1}")
print(f"Product 2: {p2}")
print(f"Product 3: {p3}\n")


dist_p1_p2_manual = math.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2 + (p1[2]-p2[2])**2)
dist_p1_p3_manual = math.sqrt((p1[0]-p3[0])**2 + (p1[1]-p3[1])**2 + (p1[2]-p3[2])**2)
dist_p2_p3_manual = math.sqrt((p2[0]-p3[0])**2 + (p2[1]-p3[1])**2 + (p2[2]-p3[2])**2)

dist_p1_p2_np = np.linalg.norm(p1 - p2)
dist_p1_p3_np = np.linalg.norm(p1 - p3)
dist_p2_p3_np = np.linalg.norm(p2 - p3)

print(f"P1 to P2: Manual = {dist_p1_p2_manual:.2f} | NumPy = {dist_p1_p2_np:.2f}")
print(f"P1 to P3: Manual = {dist_p1_p3_manual:.2f} | NumPy = {dist_p1_p3_np:.2f}")
print(f"P2 to P3: Manual = {dist_p2_p3_manual:.2f} | NumPy = {dist_p2_p3_np:.2f}")
```

```python
# TASK 2
weights = np.array([0.5, 0.3, -0.2])

five_products = df[features].head(5)

product_matrix = five_products.values

p1_values = product_matrix[0]
manual_score = (p1_values[0] * weights[0]) + (p1_values[1] * weights[1]) + (p1_values[2] * weights[2])

print(f"Manual Score (Product 1): {manual_score:.2f}\n")

scores = np.dot(product_matrix, weights)

print("--- Dot Product Scores for All 5 Products ---")
for i, score in enumerate(scores):
    print(f"Product {i+1} Score: {score:.2f}")
```
